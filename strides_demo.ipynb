```python
import torch
```

```python
t = torch.arange(0, 10).reshape(5, 2)
print(t)
```

```python
print(t.stride())
```

```python
print("Original tensor:")
print(t)
print("Shape:", t.shape)
print("Strides:", t.stride())

print("\nAfter transpose (swap rows/cols):")
t_T = t.t()
print(t_T)
print("Shape:", t_T.shape)
print("Strides:", t_T.stride())

print("\nView as 1D without copying:")
t_flat = t.view(-1)
print(t_flat)
print("Shape:", t_flat.shape)
print("Strides:", t_flat.stride())

print("\nNon-contiguous slice (every other row):")
t_slice = t[::2]
print(t_slice)
print("Shape:", t_slice.shape)
print("Strides:", t_slice.stride())
```
